### 6. In the dictionary created in Task 3 delete the key value pairs where price is less than average price of all the items in the dictionary.

```python
grocery_items = {
    "Rice": 450,
    "Olive Oil": 800,
    "Sugar": 150,
    "Almonds": 600,
    "Milk": 120
}

total_price = sum(grocery_items.values())
average_price = total_price / len(grocery_items)

keys_to_delete = [item for item, price in grocery_items.items() if price < average_price]
for key in keys_to_delete:
    del grocery_items[key]

print(grocery_items)
```

### 21.Write a python function to calculate the factorial of a given integer 10

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
result = factorial(10)
print("Factorial of 10 is:", result)

# from functools import reduce
# factorial = lambda n: reduce(lambda x, y: x * y, range(1, n + 1))
# result = factorial(10)
# print("Factorial of 10 is:", result)
```

33.	Write a function to remove duplicate elements from a list.

```python

def remove_duplicates(input_list):
    seen = set()
    output_list = []
    for item in input_list:
        if item not in seen:
            output_list.append(item)
            seen.add(item)
    return output_list


example_list = [1, 2, 3, 4, 4, 5, 5, 6]
cleaned_list = remove_duplicates(example_list)
print(cleaned_list)
```
